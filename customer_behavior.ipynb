```python
import pandas as pd
df = pd.read_csv('customer_shopping_behavior.csv')
df
```

```python
# Filter the dataset down to male customers
df_male = df[df['Gender'] == 'Male']

print(f"Male customers: {len(df_male)} of {len(df)}")

# Sorted by purchase amount, highest first
df_male.sort_values('Purchase Amount (USD)', ascending=False)
```

```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
# create a column age_group
labels = ['Young Adult','Adult','Midddle-aged','Senior']
df['age_group'] = pd.qcut(df['age'],q=4, labels = labels)
```

```python
df[['age','age_group']].head(10)
```

```python
# create column purchase_frequency_days
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```

```python
df[['discount_applied','promo_code_used']].head(10)
```

```python
(df['discount_applied'] == df['promo_code_used']).all()
```

```python
df = df.drop(columns='promo_code_used', errors='ignore')
```

```python
# Save the cleaned data to a new CSV, next to this notebook.
# index=False stops pandas writing its row numbers as an extra unnamed column.
output_path = 'customer_shopping_behavior_cleaned.csv'
df.to_csv(output_path, index=False)

print(f"Saved {len(df)} rows and {df.shape[1]} columns to {output_path}")

# Read it back to confirm it saved correctly
pd.read_csv(output_path).head()
```
